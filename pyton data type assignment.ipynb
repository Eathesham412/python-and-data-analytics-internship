```python
#Frozen set- Immutable form of sets.
newSet=frozenset([10,20,"Apples"])
print(newSet)

```

```python
#Binary types- Stores binary values
'''1)Bytes-Immutable sequence of bytes used to store binary data'''
a=bytes([10,20,100])
print(a)
'''2)Bytearray-Bytearray is mutable sequence of bytes used to store binary data'''
b=bytearray([10,57,67])
print(b)
b[0]=20
print(b)
'''3)Memoryview-It allows to access  the internal data of an object without making a copy of it'''
c=bytes([10,20,40])
d=memoryview(c)
print(d[0])
```
